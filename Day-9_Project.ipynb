1.What is the average daily social media usage (in hours) for each gender?

```python
gender_usage = (df.groupby("gender")["daily_social_media_hours"].mean().sort_values(ascending=False))
print(gender_usage)
```

```python
plt.figure(figsize=(8,5))

xy = sns.barplot(
    x=gender_usage.index,
    y=gender_usage.values,
    palette="mako"
)
for i in xy.containers:
    xy.bar_label(i, fmt='%.2f', padding=3)

plt.title("Average Daily Social Media Usage by Gender",fontsize=18,fontweight="bold")
plt.xlabel("Gender",fontsize=13,fontweight="bold")
plt.ylabel("Average Daily Social Media Usage (Hours)",fontsize=13,fontweight="bold")
plt.xticks(fontsize=10)
plt.show()
```

2.Which social media platform is the most popular among students?

```python
popular_platform = (df["platform_usage"].value_counts())
print(popular_platform)
```

```python
plt.figure(figsize=(8,5))

xy = sns.barplot(
    x=popular_platform.index,
    y=popular_platform.values,
    palette="viridis"
)
for i in xy.containers:
    xy.bar_label(i, padding=3)

plt.title("Most Popular Social Media Platforms",fontsize=18,fontweight="bold")
plt.xlabel("Social Media Platform",fontsize=13,fontweight="bold")
plt.ylabel("Number of Students",fontsize=13,fontweight="bold")
plt.xticks(rotation=90, fontsize=10)
plt.show()
```

3.How does academic performance vary with daily social media usage?

```python
correlation = df[ ["daily_social_media_hours", "academic_performance"]].corr()
print(correlation)
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="daily_social_media_hours",
    y="academic_performance",
    hue="addiction_level",
    palette="Set1",
    alpha=0.9,
    s=60
)

plt.title("Academic Performance vs Daily Social Media Usage",fontsize=18, fontweight="bold")
plt.xlabel("Daily Social Media Usage (Hours)",fontsize=13,fontweight="bold")
plt.ylabel("Academic Performance",fontsize=13,fontweight="bold")
plt.show()
```

7.What percentage of students in the dataset are classified as depressed?

```python
depressed = (df["depression_label"].value_counts(normalize=True) * 100)
print(depressed)
```

```python
labels = ["Not Depressed","Depressed"]

plt.figure(figsize=(7,7))

plt.pie(
    depressed,
    labels=labels,
    autopct="%1.1f %%",
    colors=sns.color_palette("colorblind"),
    startangle=90
)

plt.title("Percentage of Depressed Students",fontsize=18,fontweight="bold")
plt.show()
```

10.Which factors (stress, anxiety, sleep, social media usage, and physical activity) have the strongest association with depression?

```python
corr = df[
    ["stress_level","anxiety_level","sleep_hours","daily_social_media_hours","physical_activity","depression_label"]
    ].corr()

depression_corr = (
    corr["depression_label"]
    .drop("depression_label")
    .sort_values(ascending=False)
)
print(depression_corr)
```

```python
plt.figure(figsize=(8,6))

sns.barplot(
    x=depression_corr.values,
    y=depression_corr.index,
    palette="cividis"
)
plt.title("Factors Associated with Depression",fontsize=18,fontweight="bold")
plt.xlabel("Correlation",fontsize=13,fontweight="bold")
plt.ylabel("Factors",fontsize=13,fontweight="bold")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Between Factors and Depression",fontsize=18,fontweight="bold")
plt.xlabel("Variables",fontsize=13,fontweight="bold")
plt.ylabel("Variables",fontsize=13,fontweight="bold")
plt.xticks(rotation=90, fontsize=10)
plt.yticks(fontsize=10)
plt.show()
```
